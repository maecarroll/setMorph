## Delta

Delta is some feature based description of the distribution of some formative. It is some complete statement describing the cells where a occurs. The total set of delta's for a formative can be considered the maximal knowledge of a speaker about the meaning/distribution of that formative.

let $x$ be some set of features

let $y$ be some cell in $\mathrm{dist}(a)$ 

let $z$ be some cell in $\mathrm{cell}(\lambda)$ 



$ \delta_a = \{ x_1, x_2, ... x_n \} | $ 

$\forall x, \exists y, z (x \subseteq y, z) $

$ \forall y, \exists x (y \supseteq x ) $

$ \forall z \supseteq x, \exists y (z = y) $ (needs to be restricted by x)



## Simple exponence

a formatve, $a$, is said to be a *simple exponent* iff:

$\exists |\delta_a| = 1 $

$|x| = 1 | x \in \delta_a $


```python
simpleout = sm.classify_simple(ngkolmpu_df)
pd.DataFrame(simpleout)
```

### Syncretism

A formative $a$ is considered to be an example of syncretism iff:

$\neg\exists \delta_a : |\delta_a| = 1$

```python
dfsyn = sm.classify_syn(ngkolmpu_df)
pd.DataFrame(dfsyn)
```

### Cumulative exponence

A formative $a$ is considered an example of cumulative exponence iff:

$\neg\exists \delta_a : \forall x \in \delta_a, |x| = 1$

```python
dfout = sm.classify_cumulation(ngkolmpu_df)
pd.DataFrame(dfout)
```

## Unique exponence

For a given value, v:
There is only one formative for which its all its deltas must contain that value:

$\exists!a : \forall x \in \delta_a, v \in x $




```python
uniq = pd.DataFrame(sm.classify_unique(ngkolmpu_df))
uniq
```

### Allomorphy

There is more than one way to express a given value.

For a given value, $v$, we see allomorphy if:

$ \exists a, b | \forall \delta_a, \delta_b, \exists x ( v \in x $ )$


$ \{\Sigma \in \mathrm{dist}(a)| v \in \Sigma\} \neq \{\Sigma \in \mathrm{dist}(b)| v \in \Sigma\} $


```python
allo_ngkolmpu = sm.classify_allomorphy(ngkolmpu_df)

pd.DataFrame(allo_ngkolmpu)

```


### Verbose exponence

A given value is realised by two formatives simultaniously.

For a given word, $\omega$, expressing feature value, $v$:

$ \exists a, b | a, b \in W $

$ \forall \delta_a, \delta_b, \exists x ( v \in x $ )




```python
ve_ngkolmpu = sm.classify_ve(ngkolmpu_df)
pd.DataFrame(ve_ngkolmpu)
```
