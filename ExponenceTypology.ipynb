## Delta

Delta is some feature based description of the distribution of a formative, $a$. It is a complete statement describing the cells where $a$ occurs. It is a complete, restricted and non-redundant statement describing the cells where \(a\) occurs. The total set of deltas for a formative can be considered the maximal knowledge of a speaker about the meaning/distribution of that formative.

A valid **description** $\delta_a$ of a distribution, $\mathrm{dist}(a)$, is any set satisfying the following properties (there might be multiple):

1.  **Derived from the distribution**: $\delta_a$ contains only
    feature-value sets which are subsets of cells in $\mathrm{dist}(a)$:

    $$\forall x \in \delta_a, \exists y \in \mathrm{dist}(a), x \subseteq y$$

2.  **Complete**: Everything in the distribution is described

    $$\forall y \in \mathrm{dist}(a), \exists x  \in \delta_a \ (y \supseteq x )$$

3.  **Restricted**: Only cells in the distribution are described

    $$\forall x \in \delta_a, \forall z \in \mathrm{cells}, (z \supseteq x) \rightarrow (z \in \mathrm{dist}(a))$$

4.  **Non-redundant**: None of the $x$, $y$ in $\delta_a$ are subsets of
    each other.

    $$\forall x \in \delta_a, \not \exists y \in \delta_a, x \supseteq y$$

## Exponence


A formative is the exponent a set of set of values. Yet, having defined delta is still not sufficient to determine what values are properly *expressed* by *a*. We now define **exponence** as a descriptive, minimal, and informative set of valid descriptions.

.  $a$ is the **exponent** of a (unique) set of feature-value combinations which satisfies the following properties:

1.  **Descriptive**: All $x$ in $\text{exp}(a)$ are in at least one
    $\delta_a$: $$\exists \delta_a, x_i \in \delta_a$$

2.  **Minimal**: All $x$ in $\text{exp}(a)$ are equal to or subsets of
    at least one set of features in each $\delta_a$:
    $$\forall \delta_a, \exists y \in \delta_a, x_i \subseteq y$$

3.  **Informative**: For a given feature (dimension), the total sets of
    its possible values can not be in $\text{exp}$:
    $$\forall f,  \text{values(f)} \not \subseteq \text{exp}(a)$$

An empty exponence set means that the formative is part of a stem -- it fills *all* dimensions since it is always present.


## Values and their realizations


The exponence set is structured: it contains sets of combinations of values, such as **prs.ind**. Yet there is a relation between those inner values and the formatives:

The (partial) **realization** of a value $v$ is the set of all the formatives $a$ for which $v$ is in any element of $\textrm{exp}(a)$:

$$\text{real}(v) = \{ a |  \exists x \in \mathrm{exp}(a), v \in x \}$$

This function maps values to sets of formatives.

Symmetrically, the values expressed (even partially) by a formative are the union of all $x$ in $\text{exp}(f)$:

$$\text{vals}(f) = \cup_{x \in \text{exp}(f)} x$$

This function maps formatives to sets of values.


## Canonical exponence

is the case where there is a 1:1 relation between values and formatives (each value is expressed as a formative). That is to say:

$$|\mathrm{vals}(a)| = 1 \\
    \wedge \\
    |\mathrm{real}(v)| = 1$$

Thus, there are two main ways for non-canonicity to arise, depending on the violation of either statement.


# Typology based on meaning (exp and vals)

## Simple exponence (=canonical meaning)

a formatve, $a$, is said to have a *canonical meaning* (Note: we have also called this in the past *simple exponent*) iff:

$$
|\mathrm{vals}(a)| = 1
$$


A measure of interest here is the size of the *vals* set :

```python
exponents[exponents["|vals|"]==1]
```

### Syncretism


A formative $a$ is **syncretic** when it is the exponent of multiple feature-value sets:

$$|\text{exp}(a)| > 1$$

There does not exist a description of 'a', $\delta_a$, which has a cardinality of 1, i.e. one set of features.

```python
exponents[exponents["|exp|"]>1]
```

### Cumulative exponence


A formative is **cumulative** when it is the exponent of at least one set of multiple feature-value:

$$\exists x \in \text{exp}(a), |x| > 1$$

# Typology based on forms (real)

## Unique exponence

a formative, $a$, is said to have a *canonical form* (Note: we have also called this in the past *unique exponence*) iff:

$$
|\mathrm{real}(v)| = 1
$$

In other words, given a lexeme and a value, only a single formative expresses it.

The metric of interest here is the length of the reals


```python
reals[reals["|real|"] == 1]
```

### Verbose exponence


The **realization of a value in a word** ($\textrm{real}_{\omega}(v)$) as:

$$\textrm{real}_{\omega}(v) = \text{real}(v) \cap \mathrm{wordform}(\omega)$$

**verbose exponence** occurs when in a single word, a value is realized by multiple formatives. For a value $v$ and a word $\omega$:

$$| \textrm{real}_{\omega}(v) | > 1$$

```python
real_w
```

```python
real_w[real_w["|real_w|"] > 1]
```

### Allomorphy

There is more than one formative which expresses a given value and these don't co-occur.


**allomorphy** occurs there are more than one way in which $v$ is realized, across words. This could either mean that two or more distinct formatives occur in separate words, or that the value is realized through different *combinations* of a set of formatives, across words.

$$|\{\textrm{real}_{\omega}(v) |  \omega \in \lambda \}| > 1$$

```python
reals = sm.classify_allomorphy(reals, real_w)
reals
```
